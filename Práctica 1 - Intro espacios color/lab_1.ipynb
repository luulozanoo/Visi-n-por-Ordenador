### **Tarea A.3:** Conversion de imagen de color a escala de grises ``cv2.cvtColor()``.

```python
# TODO: Convert the image from color to grayscale using cv2.cvtColor().
gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY) #Ponemos el argumento para transformar de color a gris
# TODO: Check image shape → grayscale: (height, width), color: (height, width, channels).
print("Color shape:", img.shape)
print("Gray shape:", gray.shape)
# TODO: Display the image using cv2.imshow().
cv2.imshow("Nemo en gris",gray)
cv2.waitKey(1000)
cv2.destroyAllWindows()
```

### **Tarea A.4:** Cambia el tamaño de una imagen ``cv2.resize()`` a 200x200 píxeles.

```python
# TODO: Resize the image to 200×200 pixels using cv2.resize().
resized = cv2.resize(img, [200,200])
# TODO: Check the image shape
print("Resized shape:", resized.shape)
# TODO: Display the image using cv2.imshow().
cv2.imshow("Nemo resized",resized)
cv2.waitKey(1000)
cv2.destroyAllWindows()
```

### **Tarea A.5:** Recorta el centro de la imagen a un tamaño 120x160.

```python
# TODO: Crop the center of the image 'img' with an approximate size of 120x160 pixels.
h, w = img.shape[0:2]
cy, cx = [int(h/2 - 60),int(h/2+60)],[int(w/2-80),int(w/2+80)]
roi = img[cy[0]:cy[1],cx[0]:cx[1]]
# TODO: Check the image shape
print("ROI shape:",roi.shape )
# TODO: Display the image using cv2.imshow().
cv2.imshow("Nemo recortado",roi)
cv2.waitKey(1000)
cv2.destroyAllWindows()
```

### **Tarea A.6:** Rota la imagen usando ``cv2.getRotationMatrix2D`` y ``cv2.wrapAffine()``.

```python
# TODO: Rotate around the center using cv2.getRotationMatrix2D and cv2.warpAffine.
M = cv2.getRotationMatrix2D([h/2,w/2],45,1)
rotated = cv2.warpAffine(img,M,img.shape[0:2])
# TODO: Display the image using cv2.imshow().
cv2.imshow("Nemo rotado",rotated)
cv2.waitKey(1000)
cv2.destroyAllWindows()
```

### **Tarea A.7:** Guarda la imagen con ``cv2.imwrite()``

```python
# TODO: Save the image to disk as 'salida_opcionA_1.png' using cv2.imwrite().
cv2.imwrite('../data/salida_opcionA_1.png',rotated)
```

### **Tarea B.1:** Defina y ejecute el método para cargar imágenes ``load_images()``

```python
def load_images(filenames: List) -> List:
    return [cv2.imread(filename) for filename in filenames]
```

```python
# TODO Build a list containing the paths of all images in the data folder
imgs_path = []
from pathlib import Path
carpeta = Path('../data')
for archivo in carpeta.iterdir():
    if archivo.is_file() and archivo.name != ".DS_store":
        imgs_path.append(f"../data/{archivo.name}")
imgs = load_images(imgs_path)
imgs = imgs[1::]
```

### **Tarea B.2**. Defina los métodos ``show_image()`` y ``write_image()`` para visualizar y guardar imágenes

```python
# TODO Complete the method, use every argument
def show_image(img: np.array, img_name: str = "Image"):
    cv2.imshow(img_name,img)
    cv2.waitKey(10000)
    cv2.destroyAllWindows()

# TODO Complete the method, use every argument
def write_image(output_folder: str, img_name: str, img: np.array):
    img_path = os.path.join(output_folder,img_name)
    cv2.imwrite(img_path,img)
```

### **Tarea B.3:** Cambie el espacio de color de las imágenes a uno donde la crominancia y la intensidad estén separados (HSV)

```python
# TODO Get a list with ìmgs in HSV color space
hsv_imgs = []
for image in imgs:
    hsv_imgs.append(cv2.cvtColor(image,cv2.COLOR_BGR2HSV))
for image in hsv_imgs:
    show_image(image)
```

### **Tarea B.4:** Segmente los colores anaranjados

```python
# TODO Define orange color range
light_orange = (1, 190, 200)
dark_orange = (18, 255, 255)

# TODO Compute a list that contains a mask (which should segment orange colors) for every image.
orange_masks = []
for image in hsv_imgs:
    orange_masks.append(cv2.inRange(image,light_orange,dark_orange))

# TODO Compute a list that contains the result of multiplying the original image with its orange colors mask.
orange_segmented = []
for i in range(len(imgs)):
    orange_segmented.append(cv2.bitwise_and(imgs[i],imgs[i],dst = None, mask = orange_masks[i]))

# TODO Show an original image
for image in imgs:
    show_image(image)

# # TODO Show a mask
for mask in orange_masks:    
    show_image(mask)

# TODO Show a segmented image
for segmented in orange_segmented:
    show_image(segmented)
```

### **Tarea B.5:** Segmente los colores blancos.

Para detectar el rango de blancos complete la siguiente celda y ejecutela para investigar el rango de valores necesarios.

```python
# TODO Discover white color ranges
for image in hsv_imgs:
    get_hsv_color_ranges(image)

# TODO Define white color range
light_white = (0, 0, 200)       
dark_white = (179, 40, 255)
# TODO Compute a list that contains a mask (which should segment white colors) for every image.
white_masks = []
for image in hsv_imgs:
    white_masks.append(cv2.inRange(image,light_white,dark_white))

white_segmented = []
for i in range(len(imgs)):
    white_segmented.append(cv2.bitwise_and(imgs[i],imgs[i],dst = None, mask = white_masks[i]))

# TODO Show an original image
# for image in imgs:
#     show_image(image)

# TODO Show a mask
for mask in white_masks:    
    show_image(mask)

# TODO Show a segmented image
for segmented in white_segmented:
    show_image(segmented)
```

### **Tarea B.6:** Junte las máscaras para cada imagen (naranja + blanco) y segmente cada una de ellas.

```python
# TODO Join orange_masks and white_masks
fish_masks = [cv2.bitwise_or(white_masks[i], orange_masks[i]) for i in range(len(white_masks))]
    
# TODO Compute a list that contains the result of multiplying the original image with its complete mask.
fish =[]
for i in range(len(imgs)):
    fish.append(cv2.bitwise_and(imgs[i],imgs[i],dst = None, mask = fish_masks[i]))

for image in imgs:
    show_image(image)

# TODO Show a mask
for mask in fish_masks:    
    show_image(mask)

# TODO Show a segmented image
for segmented in fish:
    show_image(segmented)
```

### **Tarea B.7:** Guarde las imágenes

```python
# TODO Define your output folder and save every fish image
output_folder = "../data"
for i in range(len(fish)):
    write_image(output_folder=output_folder, img_name=f"Imagen{i}.png", img=fish[i])
```

### **Pregunta B.1:** Segmente por color el logotipo de su marca favorita: descompóngalo en al menos 3 colores.

```python
# TODO Homework
file = "../data/preg_adicional/escudo_atleti.png"
img_rgb = cv2.imread(file)
show_image(img_rgb)

img_hsv = cv2.cvtColor(img_rgb,cv2.COLOR_BGR2HSV)
show_image(img_hsv)

#Segmentaremos los colores: azul, blanco, verde, negro y rojo

# COLOR AZUL
light_blue = (100, 50, 20)
dark_blue = (140, 255, 255)
blue_mask = cv2.inRange(img_hsv, light_blue, dark_blue)
show_image(blue_mask) # 1. Máscara binaria
blue_seg = cv2.bitwise_and(img_rgb, img_rgb, mask=blue_mask)
show_image(blue_seg)  # 2. Imagen original multiplicada por la máscara

# COLOR BLANCO
light_white = (0, 0, 200)
dark_white = (179, 30, 255)
white_mask = cv2.inRange(img_hsv, light_white, dark_white)
show_image(white_mask)
white_seg = cv2.bitwise_and(img_rgb, img_rgb, mask=white_mask)
show_image(white_seg)

# COLOR VERDE
light_green = (35, 50, 50)
dark_green = (85, 255, 255)
green_mask = cv2.inRange(img_hsv, light_green, dark_green)
show_image(green_mask)
green_seg = cv2.bitwise_and(img_rgb, img_rgb, mask=green_mask)
show_image(green_seg)

# COLOR NEGRO
light_black = (0, 0, 0)
dark_black = (179, 255, 50)
black_mask = cv2.inRange(img_hsv, light_black, dark_black)
show_image(black_mask) 
black_seg = cv2.bitwise_and(img_rgb, img_rgb, mask=black_mask)
show_image(black_seg)

# COLOR ROJO -->> CIRCULAR! Segmentamos la máscara del rojo en 2
light_red1 = (0, 70, 50)
dark_red1 = (10, 255, 255)
red_mask1 = cv2.inRange(img_hsv, light_red1, dark_red1)

light_red2 = (170, 70, 50)
dark_red2 = (179, 255, 255)
red_mask2 = cv2.inRange(img_hsv, light_red2, dark_red2)

red_mask = cv2.bitwise_or(red_mask1, red_mask2)
show_image(red_mask)
red_seg = cv2.bitwise_and(img_rgb, img_rgb, mask=red_mask)
show_image(red_seg)

# CÁLCULO DE PORCENTAJES SOBRE EL TOTAL -->> Separamos porque el bitwise solo admite máscaras de 2 en 2
final_mask = cv2.bitwise_or(blue_mask, white_mask)
final_mask = cv2.bitwise_or(final_mask, green_mask)
final_mask = cv2.bitwise_or(final_mask, black_mask)
final_mask = cv2.bitwise_or(final_mask, red_mask)

final_image = cv2.bitwise_and(img_rgb,img_rgb,mask = final_mask)
show_image(final_image)


total_escudo = cv2.countNonZero(final_mask) #Esta función cuenta cuántos píxeles son distintos de 0 en la matriz

print("--- Porcentajes de color del escudo ---")
print(f"Azul: {(cv2.countNonZero(blue_mask) / total_escudo) * 100:.2f}%")
print(f"Blanco: {(cv2.countNonZero(white_mask) / total_escudo) * 100:.2f}%")
print(f"Verde: {(cv2.countNonZero(green_mask) / total_escudo) * 100:.2f}%")
print(f"Negro: {(cv2.countNonZero(black_mask) / total_escudo) * 100:.2f}%")
print(f"Rojo: {(cv2.countNonZero(red_mask) / total_escudo) * 100:.2f}%")
```
